## Question 2 : types et redondances

**Types :**
- `survived` (int64) : correct, 0 ou 1.
- `pclass` (int64) : à passer en `category`, c'est une classe (1, 2, 3) et non une quantité.
- `sex`, `who`, `embarked`, `embark_town`, `alive` (str) : correct pour du texte.
- `adult_male`, `alone` (bool) : correct.
- `class`, `deck` (category) : correct.

**Colonnes redondantes :**
- `embarked` et `embark_town` : même information (code et nom du port). Je garde `embark_town`, plus lisible.
- `class` et `pclass` : même information. Je garde `pclass`.
- `alive` et `survived` : même information. Je garde `survived`, déjà numérique, donc la moyenne fonctionne directement.
- `adult_male` : se déduit de `who` (vrai si `who` vaut `man`).

**Colonne facultative :**
- `alone` : se déduit de `sibsp` et `parch`, mais reste utile pour l'analyse.
